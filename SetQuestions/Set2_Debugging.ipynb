## Task Assignment Confusion

```python
from autogen import ConversableAgent

config = {
    "config_list": [{
        "model": "gemini-3.5-flash-lite",
        "api_type": "google",
        "api_key": "OdpLrj0vwIUnGd4PDZSSJWJ-A"
    }]
}

worker = ConversableAgent(name="WorkerAgent", llm_config=config)
planner = ConversableAgent(name="PlannerAgent", llm_config=config) 

task_plan = planner.generate_reply(messages=[{"role": "user", "content": "Plan steps to clean and organize my digital files"}])
execution = worker.generate_reply(messages=[{"role": "user", "content": task_plan["content"]}])

print("Plan:", task_plan["content"])
print("Execution:", execution["content"])
```

## Reviewer–Approver (Blog Publishing)

```python
from autogen import ConversableAgent

config = {
    "config_list": [{
        "model": "gemini-3.5-flash-lite",
        "api_type": "google",
        "api_key": "Lrj0vwIUnGd4PDZSSJWJ-A"
    }]
}

reviewer = ConversableAgent(name="ReviewerAgent", llm_config=config)
approver = ConversableAgent(name="ApproverAgent", llm_config=config)  

blog_draft = "AI is good. It helps. Many use it."

review_feedback = reviewer.generate_reply(
    messages=[{"role": "user", 
               "content": f"Please review and improve the following blog post:\n\n{blog_draft}"}]
)

approval = approver.generate_reply(
    messages=[{"role": "user", "content": f"Is this blog ready to publish? Reply YES or NO and give a short reason:\n\n{review_feedback['content']}"}]
)

print("✅ Reviewer Feedback:\n", review_feedback["content"])
print("✅ Final Approval:\n", approval["content"])
```
